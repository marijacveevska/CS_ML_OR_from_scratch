# The Basics of Numpy Arrays


```python
import numpy as np

np.random.seed(0)

x1 = np.random.randint(10,size = 6)
x2 = np.random.randint(10,size = (3,3))
x3 = np.random.randint(10,size = (3,4,5))

print(f"x1 = {x1}")
print(f"x2 = {x2}")
print(f"x3 = {x3}")
```

```python
x1[1]
```

```python
x2[2,0]
```

```python
x2[0,0] = 12
x2
```

# Array Slicing: Accessing Subarrays

x[start:stop:step]

```python
print(x1[:4])
print(x1[2:5])
```

```python
x2[:2, :3]  # two rows, three columns
```

```python
x = np.arange(4)
print("x     =", x)
print("x + 5 =", x + 5)
print("x - 5 =", x - 5)
print("x * 2 =", x * 2)
print("x / 2 =", x / 2)
print("x // 2 =", x // 2)  # floor division
print("-x     = ", -x)
print("x ** 2 = ", x ** 2)
print("x % 2  = ", x % 2)
```

```python
# Absolute value
x = np.array([-2, -1, 0, 1, 2])
abs(x)
```

```python
x = [1, 2, 3]
print("x     =", x)
print("e^x   =", np.exp(x))
print("2^x   =", np.exp2(x))
print("3^x   =", np.power(3, x))
```

## Outer products

# Example

```python
height = [189, 170, 189, 163, 183, 171, 185, 168, 173, 183, 173, 173, 175, 178, 183, 193, 178, 173, 174, 183, 183, 
          168, 170, 178, 182, 180, 183, 178, 182, 188, 175, 179, 183, 193, 182, 183, 177, 185, 188, 188, 182, 185]
heights = np.array(height)
```

```python
print(f"mean height: {np.mean(heights)}")

print(f"standard deviation: {heights.std()}")
print(f"standard deviation: {np.std(heights)}")

print("minimum: ", heights.min())
print("maximum: ", heights.max())

print("25th percentile:   ", np.percentile(heights, 25))
print("median:            ", np.median(heights))
print("75th percentile:   ", np.percentile(heights, 75))
```

```python
import matplotlib.pyplot as plt
import seaborn;
seaborn.set()

plt.hist(heights)
plt.title('Height Distribution of US Presidents')
plt.xlabel('height (cm)')
plt.ylabel('number')
```
