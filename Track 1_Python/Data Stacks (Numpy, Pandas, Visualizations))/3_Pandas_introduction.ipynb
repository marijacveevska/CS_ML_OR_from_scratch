# Pandas

![NumPy vs Pandas comparison](numpy_vs_pandas.png)

| | NumPy | Pandas |
|---|---|---|
| Core object | `ndarray` | `DataFrame` / `Series` |
| Data types | one dtype for the WHOLE array | each column can have its own dtype |
| Access | by position only (`arr[0, 1]`) | by label (`df.loc['2024-02']`) or position (`df.iloc[1]`) |
| Best for | pure numeric computation, math-heavy operations | tabular, labeled, mixed-type real-world data |
| Speed | faster — no label overhead | slightly slower — carries index/column bookkeeping |
| Built on | itself (the foundation) | built ON TOP of NumPy arrays |

```python
import pandas as pd 
df = pd.DataFrame(
    {
    'city': ['Utrecht', 'Utrecht', 'Utrecht'],
    'temp':[4.2, 6.1, 9.4],
    'pv_kw':[120, 210, 340]
    },
    index = ['2024-01','2024-02','2024-03'])

df.loc['2024-01']

#df['pv_kw'].mean()
```

## Series vs Data Frame

```python
population_dict = {'California': 38332521,
                   'Texas': 26448193,
                   'New York': 19651127,
                   'Florida': 19552860,
                   'Illinois': 12882135}
population = pd.Series(population_dict)
population
```

```python
area_dict = {'California': 423967, 'Texas': 695662, 'New York': 141297,
             'Florida': 170312, 'Illinois': 149995}
area = pd.Series(area_dict)
area
```

```python
states = pd.DataFrame({'population': population,
                       'area': area})
states
```
