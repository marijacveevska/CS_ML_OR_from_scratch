# Chapter 10 — Recursively Recurse with Recursion

## What recursion is

A function that calls **itself**. Used correctly, it solves certain 
problems — especially ones with an unknown "depth" — in surprisingly simple 
ways. Used incorrectly (no stopping condition), it causes **infinite 
recursion**, which crashes the program.

This chapter connects directly back to last chapter's stacks — Python 
itself uses a stack internally to make recursion work, which is the real 
payoff of understanding stacks first.
