## The core problem Big O solves

Saying "this algorithm takes 22 steps" isn't useful — the number of steps 
usually depends on how much data there is. Big O Notation is a standardized 
way to answer one specific question, the **"key question"**:

> If there are N data elements, how many steps will the algorithm take?

The answer goes inside the parentheses: `O(N)`, `O(1)`, `O(log N)`, etc.

![Big O notation](bigO.png)

## O(N) — Linear time

**Linear search**: in the worst case, checking every element takes as many 
steps as there are elements. For N elements → N steps → `O(N)`.

As data grows, steps grow **in direct proportion** — double the data, double 
the steps. Plotted on a graph, `O(N)` is a perfect diagonal line.

````python
def linear_search(arr, target):
    for i, value in enumerate(arr):     # up to N steps in the worst case
        if value == target:
            return i
    return -1

# O(N): steps scale directly with len(arr)
````

## O(1) — Constant time

**Reading from an array by index** (`arr[5]`) always takes exactly one step, 
no matter how large the array is. The answer to the key question has 
**nothing to do with N** — that's the whole point. Plotted on a graph, `O(1)` 
is a perfectly flat horizontal line.

`O(1)` is considered the fastest category of algorithm, since it never gets 
slower as data grows.

````python
def get_first_element(arr):
    return arr[0]     # always exactly 1 step, regardless of len(arr)

# O(1): constant time
````

## The "soul" of Big O — why O(3) is really O(1)

Big O isn't really about counting exact steps — it's about **how performance 
changes as data grows**. An algorithm that always takes exactly 3 steps, no 
matter how much data there is, is written as `O(1)`, **not** `O(3)`.

Why? Because Big O only cares about the *category* of growth behavior — 
"stays constant" vs. "grows proportionally" vs. "grows logarithmically" — not 
the exact constant. An algorithm that always takes 3 steps and one that always 
takes 1 step behave identically as data scales: neither slows down. They're 
the same *kind* of algorithm, so they get the same Big O classification.

**Key implication**: an `O(1)` algorithm that takes 1,000,000 steps will 
*eventually* be faster than an `O(N)` algorithm, once N grows large enough — 
because `O(N)`'s steps keep climbing forever, while `O(1)`'s steps never do. 
There's always some crossover point past which `O(1)` wins, no matter how 
large its constant is.

## Big O usually means worst case

Linear search is actually `O(1)` in the **best case** (the target is the very 
first element) and `O(N)` in the **worst case** (the target is last, or 
missing entirely). Unless stated otherwise, **Big O Notation refers to the 
worst case** — a deliberately pessimistic default, since knowing the worst 
case prepares you for what could actually happen.

## O(log N) — Logarithmic time

**Binary search** doesn't fit `O(1)` (its steps do increase with data) or 
`O(N)` (it's much faster than checking every element — only ~7 steps for 100 
elements). It falls in between: `O(log N)`.

**What a logarithm means**: `log₂ 8 = 3` answers "how many times do I halve 8 
until I reach 1?" → `8 / 2 / 2 / 2 = 1` → 3 halvings.

`O(log N)` means: the algorithm takes as many steps as it takes to keep 
halving N until only 1 element remains — exactly what binary search does, 
narrowing the search range in half on every step.

**Efficiency ranking so far, best to worst:**
```
O(1)  <  O(log N)  <  O(N)
```

````python
import math

# Demonstrating the log2 relationship directly
for n in [8, 16, 32, 64, 128, 256, 512, 1024]:
    print(f"N={n:>5}  O(N) steps={n:>5}  O(log N) steps={math.log2(n):.0f}")
````

This reproduces the book's comparison table — notice `O(N)` steps double 
every row, while `O(log N)` steps only go up by 1 each time N doubles. This 
gap becomes enormous as N grows large.

````python
# Attempt each one here, then state your Big O answer as a comment

def is_leap_year(year):
    return (year % 100 == 0 and year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)
````

2. Use Big O Notation to describe the time complexity of the following 
   function that sums up all the numbers from a given array:

````python
def array_sum(array):
    total = 0
    for x in array:
        total += x
    return total
````

it is O(n) because if the array increases more sums will happen. The number of total += x operations grows in direct proportion to len(array). Double the array, double the steps. That's the defining signature of O(N).

3. Use Big O Notation to describe the time complexity of the following function

````python
def chessboard_space(number_of_grains):
    chessboard_spaces = 1
    placed_grains = 1
    while placed_grains < number_of_grains:
        placed_grains *= 2
        chessboard_spaces += 1
    return chessboard_spaces
````

O(logn).\

Say number_of_grains = 16. Trace it:

- Start: placed_grains = 1
- Iteration 1: placed_grains = 2
- Iteration 2: placed_grains = 4
- Iteration 3: placed_grains = 8
- Iteration 4: placed_grains = 16 → loop stops (no longer < 16)

4 iterations to go from 1 to 16. 

log₂ 16 = 4

4. What the input is: array is a list of strings (e.g. ["apple", "banana", "avocado", "cherry"]). The function loops through it once, checking each string individually, and builds up a new list containing only the strings that start with "a".

````python
def select_a_strings(array):
    new_array = []
    for x in array:
        if x.startswith("a"):
            new_array.append(x)
    return new_array
````

O(n). One loop through the array and taking the words that start with "a"

5. What the input is: array is a list of numbers that's already sorted (this matters — the function assumes it, it doesn't sort anything itself). The function computes the median (the middle value) by going straight to the middle index (or two middle indices, if the array has an even length) and reading the value(s) there directly — no searching, no looping.

````python
def median(array):
    middle = len(array) // 2
    if len(array) % 2 == 0:
        return (array[middle - 1] + array[middle]) / 2
    else:
        return array[middle]
````

The function does at most two direct index lookups (array[middle - 1] and array[middle], or just array[middle]) plus one division/modulo check. None of that depends on len(array) — whether the array has 10 elements or 10 million, you're still doing exactly the same fixed handful of operations: compute middle (one division), check even/odd (one modulo), read one or two specific indices.
